```python
import pandas as pd

df = pd.read_csv(
    "../data/raw/Ames/AmesHousing.csv"
)

print(df.shape)

df.head()
```

```python
df.info()
```

```python
missing = (
    df.isnull()
      .sum()
      .sort_values(ascending=False)
)

missing.head(20)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,6))

sns.histplot(
    df["SalePrice"],
    bins=50,
    kde=True
)

plt.title("SalePrice Distribution")
plt.xlabel("Sale Price")
plt.ylabel("Count")

plt.show()
```

```python
corr = (
    df.corr(numeric_only=True)
      ["SalePrice"]
      .sort_values(ascending=False)
)

corr.head(15)
```
